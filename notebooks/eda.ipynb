# **Exploratory Data Analysis**

## **0. Imports**

```python
import numpy as np 
import pandas as pd
```

## **1. Load Data**

```python
customer = pd.read_csv('../data/raw/raw_customers.csv')
item = pd.read_csv('../data/raw/raw_order_items.csv')
order = pd.read_csv('../data/raw/raw_orders.csv')
product = pd.read_csv('../data/raw/raw_products.csv')
store = pd.read_csv('../data/raw/raw_stores.csv')
```

## **2. EDA: Stores**

```python
print("Store dataset shape:", store.shape)
print("Store columns and NULL vales:")
print(store.isnull().sum())
display(store.head(10))
```

### **Findings (Store Table):**
- 10 rows, 5 columns
- No missing Values
- Inconsistent city casing
- No primary keys present

## **3. EDA: Products**


```python
print("Products dataset shape:", product.shape)
print("\nProducts columns and NULL values:")
print(product.isnull().sum())

print("\nTrailing spaces per column:")
for col in product.select_dtypes(include='object').columns:
    count = product[col].str.endswith(' ').sum()
    print(f"  {col}: {count}")

print("\nUnique categories:", product['category'].nunique())
display(product.head())
```

### **Findings (Product Table):**
- 145 rows, 5 columns
- 29 missing values (13 in price, 16 in stock quantity)
- 25 products with trailing spaces in product_name 
- 30 unique category values for 6 actual categories:
    - Inconsistent casing: Electronics, electronics, ELECTRONICS, electronic, Elektronica
    - Dutch translations mixed in: Voedsel, Wonen, Kleding, Schoonheid, Sport
    - Plural/singular variations: Food vs Foods, Sport vs Sports
    - Alternative names: Personal Care, Home & Living, Sporting Goods, clothes
- No primary key present 

## **4. EDA: Customers**

```python
print("Customer dataset shape:", customer.shape)
print("\nCustomer columns and NULL values:")
print(customer.isnull().sum())

print("\nTrailing spaces per column:")
for col in customer.select_dtypes(include='object').columns:
    count = customer[col].str.endswith(' ').sum()
    print(f"  {col}: {count}")

print("\nDuplicate rows:", customer.duplicated().sum())

print("\nUnique values per categorical column:")
for col in customer.select_dtypes(include='object').columns:
    print(f"\n{col} — {customer[col].nunique()} unique values")
    print(customer[col].value_counts().head(10))

display(customer.head())
```

### **Findings: (Customer Table)**
- 640 rows, 7 columns
- 77 missing emails 
- 42 missing phone numbers 
- No trailing spaces in any column
- 40 exact duplicate rows
- 7 unique country variations for 1 actual country:
    - Inconsistent casing: Netherlands, NETHERLANDS
    - Abbreviations: NL, NLD, nl
    - Alternative names: The Netherlands, nederland
- city and date_of_birth appear consistent, no cleaning needed
- Duplicate emails detected — likely caused by duplicate rows
- No primary key present — needs to be generated in cleaning

## **5. EDA: Orders**


```python
print("Orders dataset shape:", order.shape)
print("\nOrders columns and NULL values:")
print(order.isnull().sum())

print("\nTrailing spaces per column:")
for col in order.select_dtypes(include='object').columns:
    count = order[col].str.endswith(' ').sum()
    print(f"  {col}: {count}")

print("\nDuplicate rows:", order.duplicated().sum())

print("\nUnique values per categorical column:")
for col in order.select_dtypes(include='object').columns:
    print(f"\n{col} — {order[col].nunique()} unique values")
    print(order[col].value_counts().head(10))

display(order.head())
```

## **6. EDA: Order Items**


```python
print("Order items dataset shape:", item.shape)
print("\nOrder items columns and NULL values:")
print(item.isnull().sum())

print("\nTrailing spaces per column:")
for col in item.select_dtypes(include='object').columns:
    count = item[col].str.endswith(' ').sum()
    print(f"  {col}: {count}")

print("\nDuplicate rows:", item.duplicated().sum())

print("\nUnique values per categorical column:")
for col in item.select_dtypes(include='object').columns:
    print(f"\n{col} — {item[col].nunique()} unique values")
    print(item[col].value_counts().head(10))

print("\nWrong line totals:")
item['expected_total'] = item['unit_price'] * item['quantity']
print(item[item['line_total'] != item['expected_total']].shape[0], "incorrect line totals")

display(item.head())
```

### **Findings: (Order Items Table)**
- 3000 rows, 5 columns
- 265 missing quantities 
- 236 missing unit prices 
- 481 missing line totals  
- No trailing spaces in any column

- order_reference inconsistent, mix of ORD_XXXX format and customer names

- No primary key or foreign keys present
